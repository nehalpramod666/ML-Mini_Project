## Inference validation

The final Q4_K_M model loaded successfully with `llama-cli`.

Smoke tests passed for:
- ACCEPT / none
- low_contrast
- sensor_noise
- high_contrast
- motion_blur

Motion-blur test:

`blur=85, contrast=52, noise=20, state=REJECT, dominant_defect=motion_blur`

The model identified motion blur, preserved the supplied blur score, and recommended repositioning and repeating the image.


## Known observations

The model is functionally usable for the prototype.

Known output-quality issues:
- Blur values around 85–90 are sometimes described as “moderate” instead of “severe”.
- Sensor-noise responses can contain extra specific recommendations.

These are model-output quality observations, not GGUF conversion or quantization failures.
