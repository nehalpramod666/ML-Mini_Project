```python
from transformers import TrainingArguments
from trl import SFTTrainer
import torch
import os

CHECKPOINT_DIR = "/content/drive/MyDrive/clinical-gatekeeper-checkpoints"

checkpoint_dirs = sorted([
    os.path.join(CHECKPOINT_DIR, d)
    for d in os.listdir(CHECKPOINT_DIR)
    if d.startswith("checkpoint-")
    and os.path.isdir(os.path.join(CHECKPOINT_DIR, d))
])

resume = len(checkpoint_dirs) > 0

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset_text["train"],
    eval_dataset=dataset_text["val"],
    dataset_text_field="text",
    max_seq_length=2048,
    args=TrainingArguments(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,

        num_train_epochs=3,

        warmup_steps=51,
        learning_rate=2e-4,
        lr_scheduler_type="cosine",

        fp16=not torch.cuda.is_bf16_supported(),
        bf16=torch.cuda.is_bf16_supported(),

        logging_steps=10,

        eval_strategy="steps",
        eval_steps=100,

        save_strategy="steps",
        save_steps=200,
        save_total_limit=3,

        optim="adamw_8bit",

        output_dir=CHECKPOINT_DIR,
        report_to="none",
    ),
)

print("Trainer created successfully.")
```

```python
import json
import torch

SYSTEM_PROMPT = (
    "You are a clinical imaging quality assistant. Given DICOM quality metrics "
    "(blur, contrast, noise scores, overall state, and dominant defect), explain "
    "the defect and give a concise, actionable retake instruction."
)

def generate(metrics_dict):
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": json.dumps(metrics_dict)},
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt",
        return_dict=True,
    )

    inputs = {
        key: value.to("cuda")
        for key, value in inputs.items()
    }

    outputs = model.generate(
        input_ids=inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_new_tokens=80,
        use_cache=True,
        do_sample=False,
    )

    generated_ids = outputs[0][inputs["input_ids"].shape[1]:]

    return tokenizer.decode(
        generated_ids,
        skip_special_tokens=True,
    ).strip()
```

```python
def extract_val_example(row):
    messages = row["messages"]

    user_message = next(
        m["content"] for m in messages if m["role"] == "user"
    )

    expected = next(
        m["content"] for m in messages if m["role"] == "assistant"
    )

    metrics = json.loads(user_message)

    return metrics, expected
```

```python
import random

random.seed(42)
sample_rows = random.sample(val_rows, 15)

for i, row in enumerate(sample_rows, 1):
    metrics, expected = extract_val_example(row)
    generated = generate(metrics)

    print("\n" + "=" * 80)
    print(f"VAL EXAMPLE {i}/15")
    print("INPUT:   ", metrics)
    print("EXPECTED:", expected)
    print("GOT:     ", generated)
```
