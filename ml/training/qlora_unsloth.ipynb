```python
max_seq_length = 2048
dtype = None
load_in_4bit = True

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Llama-3.2-3B-Instruct-bnb-4bit",
    max_seq_length=max_seq_length,
    dtype=dtype,
    load_in_4bit=load_in_4bit,
)

print("Model loaded successfully!")
```

```python
FastLanguageModel.for_inference(model)

print("Inference mode enabled")
```

```python
prompt = (
    "Image rejected due to blur=85, contrast=40, state=REJECT. "
    "Explain the defect and give a one-sentence retake instruction. "
    "Use only the information provided. Do not invent measurements, "
    "focus values, or other technical parameters."
)

inputs = tokenizer(
    [prompt],
    return_tensors="pt"
).to("cuda")

outputs = model.generate(
    **inputs,
    max_new_tokens=100,
    use_cache=True
)

response = tokenizer.batch_decode(
    outputs,
    skip_special_tokens=True
)[0]

print(response)
```

```python
from datasets import load_dataset

dataset = load_dataset(
    "json",
    data_files={
        "train": "/content/train_chat.jsonl",
        "val": "/content/val_chat.jsonl",
    },
)

print(dataset)
print(dataset["train"][0])
```

```python
def format_batch(examples):
    texts = []

    for messages in examples["messages"]:
        text = tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=False,
        )
        texts.append(text)

    return {"text": texts}


dataset_text = dataset.map(
    format_batch,
    batched=True,
    remove_columns=["messages"],
)

print(dataset_text)
print(dataset_text["train"][0]["text"])
```

```python
from unsloth import FastLanguageModel

model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=[
        "q_proj",
        "k_proj",
        "v_proj",
        "o_proj",
        "gate_proj",
        "up_proj",
        "down_proj",
    ],
    lora_alpha=16,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=42,
)

model.print_trainable_parameters()
```

```python
import torch
from trl import SFTTrainer, SFTConfig

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

training_args = SFTConfig(
    output_dir="outputs_smoke_test_eval",
    max_steps=30,

    per_device_train_batch_size=2,
    per_device_eval_batch_size=2,
    gradient_accumulation_steps=4,

    warmup_steps=5,
    learning_rate=2e-4,

    logging_steps=1,
    logging_first_step=True,

    eval_strategy="steps",
    eval_steps=10,

    fp16=not torch.cuda.is_bf16_supported(),
    bf16=torch.cuda.is_bf16_supported(),

    optim="adamw_8bit",

    max_length=2048,
    dataset_text_field="text",

    report_to="none",
    seed=42,
)

trainer = SFTTrainer(
    model=model,
    args=training_args,
    train_dataset=dataset_text["train"],
    eval_dataset=dataset_text["val"],
    processing_class=tokenizer,
)

print("SFTTrainer configured successfully.")
```

```python
print("Train rows:", len(trainer.train_dataset))
print("Eval rows:", len(trainer.eval_dataset))
print("Max length:", trainer.args.max_length)
print("Max steps:", trainer.args.max_steps)
```

```python
trainer_stats = trainer.train()
```

```python
from unsloth import FastLanguageModel

FastLanguageModel.for_inference(model)

test_input = (
    '{"blur": 85.2, "contrast": 40.1, "noise": 12.4, '
    '"state": "REJECT", "dominant_defect": "motion_blur"}'
)

messages = [
    {
        "role": "system",
        "content": (
            "You are a clinical imaging quality assistant. "
            "Given DICOM quality metrics (blur, contrast, noise scores, "
            "overall state, and dominant defect), explain the defect and "
            "give a concise, actionable retake instruction."
        ),
    },
    {
        "role": "user",
        "content": test_input,
    },
]

inputs = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
    return_dict=True,
).to("cuda")

outputs = model.generate(
    input_ids=inputs["input_ids"],
    attention_mask=inputs["attention_mask"],
    max_new_tokens=80,
    use_cache=True,
)

print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
