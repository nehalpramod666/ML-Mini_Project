```python
from unsloth import FastLanguageModel
import torch

print("Unsloth imported successfully")
```

```python
max_seq_length = 2048
dtype = None
load_in_4bit = True

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Llama-3.2-3B-Instruct-bnb-4bit",
    max_seq_length=max_seq_length,
    dtype=dtype,
    load_in_4bit=load_in_4bit,
)

print("Modle loaded successfully!")
```

```python
FastLanguageModel.for_inference(model)

print("Inference mode enabled")
```

```python
prompt = (
    "Image rejected due to blur=85, contrast=40, state=REJECT. "
    "Explain the defect and give a one-sentence retake instruction. "
    "Use only the information provided. Do not invent measurements, "
    "focus values, or other technical parameters."
)

inputs = tokenizer(
    [prompt],
    return_tensors="pt"
).to("cuda")

outputs = model.generate(
    **inputs,
    max_new_tokens=100,
    use_cache=True
)

response = tokenizer.batch_decode(
    outputs,
    skip_special_tokens=True
)[0]

print(response)
```

```python
print("===== M1 DAY 1 SMOKE TEST =====")
print("GPU:", torch.cuda.get_device_name(0))
print("CUDA available:", torch.cuda.is_available())
print("VRAM (GB):", round(
    torch.cuda.get_device_properties(0).total_memory / 1024**3, 2
))
print("Model: unsloth/Llama-3.2-3B-Instruct-bnb-4bit")
print("4-bit loading: True")
print()
print("GENERATED OUTPUT:")
print(response)
```
